```python
def calculate_reading_speed(pages, minute):
  if minute >= 1:
    return pages/minute
  else:
    return 0

def classify_speed(reading_speed):
  if reading_speed < 1:
    return ("Developing")
  elif reading_speed > 2:
    return("Fast")
  else:
    return("steady")

total_pages = 0
total_reading_time = 0


reading_sessions = int(input("Enter reading sessions: "))

for i in range(reading_sessions):
  pages = float(input("\nEnter pages read: "))
  minutes = float(input("Enter time spent: "))

  total_pages += pages
  total_reading_time += minutes

  reading_speed = calculate_reading_speed(pages, minutes)
  classification = classify_speed(reading_speed)

  print("Total reading time: ", reading_speed)
  print("Classification: ", classification)

total_reading_speed = total_pages/total_reading_time

print("\nFINAL READING SUMMARY")
print("Total pages read: ", total_pages, "pages")
print("Total reading time: ", total_reading_time, "minutes")
print("Total reading speed: ", total_reading_speed, "minutes/page")
```
